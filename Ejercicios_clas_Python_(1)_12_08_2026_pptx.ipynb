##**Diapositiva 6**

#####-Evaluan una condición

#####-Permiten tomar decisiones en el programa

#####**Idea:** Primero se evalúa la condición y luego se ejecuta la ruta correspondiente

#####-Primero se evalúa la condición mas especifica y luego la mas general

```python
nota = 4.7
if nota >= 4.5:
  print ("Excelente")
elif nota >= 3.0:
  print ("Aprobado")
```

##**Diapositiva 8**

###**Evaluación de arriba hacia abajo**
#####-Python evalúa las condiciones de arriba hacia abajo
#####-Se ejecuta solo la primera condición verdadera
#####-Else captura los demas casos

#####**Idea:** El orden de lectura importa

```python
nota = 4.2
if nota >= 4.5:
  print ("Excelente")
elif nota >= 4.0:
  print ("Muy buen desempeño")
elif nota >= 3.0:
  print ("Aprobado")
else:
  print ("No aprobado")
```

##**Diapositiva 11**

##**Diapositiva 12**

###**Combinar for + if**
#####Recorrer cada elemento, evaluar y actuar

#####La repetición controla el recorrido; la condición filtra que mostrar.

```python
for numero in range (1,11):
  if numero % 2 ==0:
    print (numero,"es par")
```

##**Diapositiva 13**

###**Estruturas repetitivas: while**
#####-Repite mientras una condición sea verdadera.
#####-Es útil no conocemos de antemano el número de iteraciones.
#####-Requiere actualizar la variable de control

```python
numero = 1
while numero <= 5:
  print(numero)
  numero= numero + 1
```

##**Diapositiva 14**

###**While aplicado: No sabemos cuántos intentos habrá**
#####-No conocemos cuántas veces se repetirá el ciclo.

```python
Clave_correcta = "python 123"
Clave = ""

while Clave != Clave_correcta:
  Clave = input ("Ingrese la contraseña:")

print ("Acceso autorizado")
```
