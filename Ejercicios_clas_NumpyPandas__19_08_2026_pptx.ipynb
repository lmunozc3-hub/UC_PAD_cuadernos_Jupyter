##**Diapositiva 2**

###**Del ciclo al arreglo: por que Numpy**
#####Una lista de Python guarda objetos sueltos; un arreglo de NumPy guarda números del mismo tipo, uno junto a otro.

#####Con listas de Python

```python
precios = [24000,28000,22000,30000,8000]

con_iva=[]
for p in precios:
  con_iva.append (p* 1.19)

print(con_iva)
print(precios)
```

#####Con arreglos Numpy

```python
import numpy as np

precios = np.array(
    [24000, 28000, 22000, 30000, 8000]
)

con_iva= precios * 1.19

print(con_iva)
print(precios)
```

##**Diapositiva 4**

###**Operaciones vectorizadas y broadcasting**
#####Dos arreglos de la misma forma operan elemento por elemento; un escalar se estira para alcanzar a todos.

```python
cantidades = np.array([2, 1, 3, 2, 1])
precios = np.array([24000, 28000, 22000, 30000, 8000])

ingresos = cantidades * precios
con_descuento = precios * 0.9
total = ingresos.sum()

print(ingresos)
print(con_descuento)
print(total)
```

##**Diapositiva 5**

###**Indexación, rebanado y máscaras**
#####La misma sintaxis de las listas, más una capacidad nueva: seleccionar con una condición.

```python
precios = np.array([24000, 28000, 22000, 30000, 8000])

precios[0]
precios[-1]
precios[1:4]

mascara = precios > 20000
precios[mascara]
precios[mascara].size

categoria = np.where(precios > 25000, "alto", "normal")

print(precios[0])
print(precios[-1])
print(precios[1:4])

print(mascara)
print(precios[mascara])
print(precios[mascara].size)
print(categoria.dtype)
print(categoria)
```

```python
#mascaras
# &: para Y; |para la condición "o"

mascara_2 = (precios > 5000) & (precios < 26000)
mascara_2
print(f"Resultado de la mascara #2:{mascara_2}")

mascara_3 = (precios > 5000) | (precios < 26000)
mascara_3
print(f"Resultado de la mascara #3:{mascara_3}")

print(precios)
print(precios[mascara_2])
print(precios[mascara_3])
```

##**Diapositiva 6**

###**Agregar y resumir**
#####Las funciones de resumen colapsan el arreglo en un número, o en un número por fila o por columna.

```python
ingresos = np.array(
    [48000, 28000, 66000, 60000, 8000])

ingresos.sum()     # 210000
ingresos.mean()    # 42000.0
ingresos.max()     # 66000
ingresos.std()     # 21391.59
```

```python
matriz = np.array([[48000, 28000],
                   [66000, 60000],
                   [ 8000, 12000]])

matriz.sum()          # 222000
matriz.sum(axis=0)    # [122000 100000]
matriz.sum(axis=1)    # [76000 126000 20000]
```
