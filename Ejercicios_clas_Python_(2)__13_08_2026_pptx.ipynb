##**Diapositiva 8**

###**Metodos para modificar listas**
####La lista cambia despues de crearse

######*-append() agrega al final*
######*-insert() agrega en una posición*
######*-remove() borra por valor*
######*-pop() borra por indice y devuelve*

######**Idea clave:** Estos métodos modifican la lista original, no crean una copia.

```python
compras= ["Pan","Leche"]

compras.append("Huevos")
print(compras)

compras.insert(0,"Cafe")
compras.remove("Leche")

ultimo = compras.pop()
print(ultimo)
print(compras)
```

##**Diapositiva 9**

###**Recorrer una lista con for**
####Tres manera de pasar por cada elemento

```python
#1. Por elemento
#Cuando solo necesito el valor.

precios= [1200, 950, 3400]

for p in precios:
  print(p)
```

```python
#2. Por indice
#Cuando necesito la posición

precios= [1200, 950, 3400]

for i in range(len(precios)):
  print(i, precios[i])
```

```python
#3. Con enumerate
#Cuando necesito ambos a la vez

precios= [1200, 950, 3400]

for i,p in enumerate(precios):
  print(i,p)

```

##**Diapositiva 14**

###**Recorrer un diccionario**
####Tres vistas sobre los mismos datos

######*.keys()---->Recorre solo las claves.*
######*.values()-->Recorre solo los valores.*
######*.items()--->Recorre clave y valor a la vez.*

```python
stock = {"pan":12, "leche":5, "cafe":8}

for clave in stock:
  print(clave)
```

```python
stock = {"pan":12, "leche":5, "cafe":8}

for valor in stock.values():
  print(valor)
```

```python
stock = {"pan":12, "leche":5, "cafe":8}

for clave,valor in stock.items():
  print(clave,"->",valor)
```

##**Diapositiva 17**

###**Ejercicio integrador**
####Inventario con listas y diccionarios

######**Enunciado:** *Registre 3 productos con nombre y precio, calcule el total y muestre el productomás caro.*

```python
productos=[
    {"nombre":"pan","precio": 3500},
    {"nombre":"leche","precio": 4200},
    {"nombre":"cafe","precio": 12800},
]

total = 0
caro = productos[0]

for p in productos:
  total = total + p["precio"]
  if p ["precio"] > caro["precio"]:
       caro = p

print("Total:",total)
print("Mas caro:", caro["nombre"])
```
