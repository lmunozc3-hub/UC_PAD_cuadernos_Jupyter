####Dos maneras de resolver el mismo problema

#####**Sin función**

##**Diapositiva 11**

###**Por posición o por nombre**
####Dos formas de pasar los argumentos

```python
def perfil(nombre, edad, ciudad):
  print(nombre, edad, ciudad)

# Por posicion: El orden decide el destino
perfil("Ana", 20,"Bogota")

# Por nombre (keyword): El orden deja de importar.
perfil(edad=20, nombre="Ana",
       ciudad="Bogota")

# Mezcla: las posicionales primero. Las posicionales van primero.
perfil("Ana", ciudad="Bogota",
       edad=20)
```

##**Diapositiva 17**

###**Ejercicio integrador**
####Notas y aprobación con funciones

#####**Enunciado**: Cree una función que calcule el promedio de una lista de notas y otra que indique si ese promedio aprueba.

```python
#Datos de entrada

notas = [3.5, 4.2, 2.8]
nota_minima = 3.0

def promedio(notas):
  total = 0
  for n in notas:
    total = total + n
    return total / len(notas)

def aprueba(prom, minimo=3.0):
  return prom >= minimo

notas = [3.5, 4.2, 2.8]
prom = promedio(notas)

print("Promedio:", prom)
if aprueba(prom):
  print("Aprobado")
else:
  print("No aprobado")
```
