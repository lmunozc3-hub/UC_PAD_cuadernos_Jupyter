##**F.  Filtrar, derivar y agrupar**

#### 26.  Filtrar con dos condiciones
##### Almuerzos de la sede Centro. Cada comparación va entre paréntesis.

```python
mascara = (
    (df_ventas["sede"] == "Centro")
    & (df_ventas["categoria"] == "Almuerzo")
)

seleccion = df_ventas.loc[mascara].copy()

print(seleccion.shape[0], seleccion["cantidad"].sum())
```

#### 27.  Columnas derivadas  
##### Cree las tres columnas que permiten hablar de rentabilidad.

```python
df_ventas["ingreso"] = df_ventas["cantidad"] * df_ventas["precio_unitario"]
df_ventas["costo_total"] = df_ventas["cantidad"] * df_ventas["costo_unitario"]
df_ventas["margen_bruto"] = df_ventas["ingreso"] - df_ventas["costo_total"]

print(df_ventas["ingreso"].sum(), df_ventas["margen_bruto"].sum())
```

```python
print(df_ventas[["cantidad", "precio_unitario", "costo_unitario"]].dtypes)
```

#### 28.  Indicadores globales
##### El porcentaje se calcula después de sumar, no antes.

```python
ingreso_total = df_ventas["ingreso"].sum()
margen_total = df_ventas["margen_bruto"].sum()

porcentaje_global = round(margen_total / ingreso_total * 100, 2)

ticket_promedio = round(ingreso_total / len(df_ventas), 2)

print(porcentaje_global, ticket_promedio)
```

#### 29.  Agrupar por producto
##### Construya la tabla de indicadores y observe el resultado antes de responder.

```python
kpi = (
    df_ventas
    .groupby("producto")
    .agg(
        ingreso=("ingreso", "sum"),
        margen=("margen_bruto", "sum"),
    )
    .reset_index()
)

kpi["porcentaje_margen"] = (kpi["margen"] / kpi["ingreso"] * 100).round(2)
print(kpi)
```

#### 30.  Reto integrador: el resumen ejecutivo
##### Encapsule el análisis en una función que reciba el DataFrame y no dependa de ninguna variable global. Después aplíquela solo a la sede Centro.

```python
def resumen_ejecutivo(df):
    ingreso = df["ingreso"].sum()
    margen = df["margen_bruto"].sum()
    return {
        "total_ingresos": ingreso,
        "margen_bruto": margen,
        "porcentaje_margen": round(margen / ingreso * 100, 2),
        "ticket_promedio": round(ingreso / len(df), 2),
    }

print(resumen_ejecutivo(df_ventas))

solo_centro = df_ventas[df_ventas["sede"] == "Centro"].copy()
print(resumen_ejecutivo(solo_centro))
```
